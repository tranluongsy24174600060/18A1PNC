```python
#bai 6.1
import threading

def task():
    print("Luồng đang chạy:", threading.current_thread().name)

threads = []

for i in range(3):
    t = threading.Thread(target=task, name=f"Thread-{i+1}")
    threads.append(t)
    t.start()

for t in threads:
    t.join()
```

```python
#bai 6.2
import threading
import time

def download(file):
    print(f"Bắt đầu tải {file}")
    time.sleep(2)
    print(f"Hoàn thành tải {file}")

files = ["file1", "file2", "file3"]
threads = []

for f in files:
    t = threading.Thread(target=download, args=(f,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()
```

```python
#bai 6.3
import threading

def even():
    for i in range(30, 51):
        if i % 2 == 0:
            print("Chẵn:", i)

def odd():
    for i in range(30, 51):
        if i % 2 != 0:
            print("Lẻ:", i)

t1 = threading.Thread(target=even)
t2 = threading.Thread(target=odd)

t1.start()
t2.start()

t1.join()
t2.join()
```

```python
#bai 6.4
import threading

def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    print(f"Giai thừa của {n} là {result}")

numbers = [5, 6, 7]
threads = []

for n in numbers:
    t = threading.Thread(target=factorial, args=(n,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()
```

```python
#bai 6.7
import threading
import time

def website(name, delay, times):
    print(f"Starting {name}")
    for _ in range(times):
        print(f"{name}: Web {time.ctime()}")
        time.sleep(delay)
    print(f"Exiting {name}")

print("Starting Main Thread")

t1 = threading.Thread(target=website, args=("Google", 1, 4))
t2 = threading.Thread(target=website, args=("Yahoo", 2, 3))
t3 = threading.Thread(target=website, args=("Facebook", 3, 3))

t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()

print("Exiting Main Thread")
```

```python
#bai 6.8
import threading
import random

# Tạo list ngẫu nhiên
n = 40
lst = [random.randint(0, 10) for _ in range(n)]
print("List:", lst)

num_threads = 4
sub_size = n // num_threads

total_sum = 0
lock = threading.Lock()

def sum_sublist(sublist, index):
    global total_sum
    sub_sum = sum(sublist)
    print(f"Thread {index}: {sublist} -> Tổng = {sub_sum}")
    with lock:
        total_sum += sub_sum

threads = []

for i in range(num_threads):
    start = i * sub_size
    end = (i + 1) * sub_size if i != num_threads - 1 else n
    t = threading.Thread(target=sum_sublist, args=(lst[start:end], i + 1))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print("Tổng list:", total_sum)
```

```python
#bai 6.10
import threading
import random

# Tạo list ngẫu nhiên
n = 30
lst = [random.randint(0, 100) for _ in range(n)]
print("List:", lst)

num_threads = 3
sub_size = n // num_threads

max_values = []
lock = threading.Lock()

def find_max(sublist, index):
    sub_max = max(sublist)
    print(f"Thread {index}: {sublist} -> Max = {sub_max}")
    with lock:
        max_values.append(sub_max)

threads = []

for i in range(num_threads):
    start = i * sub_size
    end = (i + 1) * sub_size if i != num_threads - 1 else n
    t = threading.Thread(target=find_max, args=(lst[start:end], i + 1))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print("Giá trị lớn nhất trong list:", max(max_values))
```

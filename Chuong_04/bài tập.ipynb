```python
# bài 4.4
ser1 = pd.Series([1, 2, 3, 4, 5])
ser2 = pd.Series([4, 5, 6, 7, 8])
# tu ser1 xoa cac muc co trong ser2
ser_difference = ser1[~ser1.isin(ser2)]
print("\nSeries sau khi loại bỏ các mục có trong ser2 từ ser1:")
print(ser_difference)
# lay tat ca cac muc co trong ser1 va ser2 nhung khong nam trong ca hai
ser_symmetric_difference = pd.concat([ser1[~ser1.isin(ser2)], ser2[~ser2.isin(ser1)]])
print("\nSeries chứa tất cả các mục không nằm trong cả hai Series:")
print(ser_symmetric_difference)
```

```python
# bài 4.5
ser = pd.Series(np.random.normal(10, 5, 25))
# gia tri toi thieu (minimum) cua ser
min_value = ser.min()
print("\nGiá trị tối thiểu của Series:")
print(min_value)
# phan centile thu 25 (25th percentile) cua ser
percentile_25 = ser.quantile(0.25)
print("\nPhần centile thứ 25 của Series:")
print(percentile_25)
# trung vi (median) cua ser
median_value = ser.median()
print("\nTrung vị của Series:")
print(median_value)
# phan centile thu 75 (75th percentile) cua ser
percentile_75 = ser.quantile(0.75)
print("\nPhần centile thứ 75 của Series:")
print(percentile_75)
# gia tri toi da (maximum) cua ser
max_value = ser.max()
print("\nGiá trị tối đa của Series:")
print(max_value)
```

```python
# bài 4.7
np.random.RandomState(100)
ser = pd.Series(np.random.randint(1, 5, [12]))
# tim 2 phantu xuat hien nhieu nhat trong ser
top_2 = ser.value_counts().nlargest(2)  
print("\nHai phần tử xuất hiện nhiều nhất trong Series:")
print(top_2)
# thay cac phan tu khong phai 2 phan tu tren bang gia tri other
ser_replaced = ser.apply(lambda x: x if x in top_2.index else 'Other')
print("\nSeries sau khi thay các phần tử không phải hai phần tử trên bằng 'Other':")
print(ser_replaced)
```

```python
# bài 4.8
ser = pd.Series(np.random.random(20))
# chia ser thanh 10 phan vi bang nhau
bins = pd.qcut(ser, 10) 
print("\nCác phần bin sau khi chia Series thành 10 phần vị bằng nhau:")
print(bins)
# thay cac gia tri trong ser bang ten phan vi tuong ung
ser_binned = pd.qcut(ser, 10, labels=[f'Bin{i+1}' for i in range(10)])
print("\nSeries sau khi thay các giá trị bằng tên phần vị tương ứng:")
print(ser_binned)
```

```python
# bài 4.19
import pandas as pd
import numpy as np
np.random.seed(100)
fruit = pd.Series(np.random.choice(['apple', 'banana', 'carrot'], 10))
weights = pd.Series(np.linspace(1, 10, 10))

print("Series fruit:")
print(fruit)
print("\nSeries weights:")
print(weights)
print("\n" + "="*40 + "\n")

df = pd.DataFrame({'Fruit': fruit, 'Weight': weights})

average_weight_by_fruit = df.groupby('Fruit')['Weight'].mean()

print("Kết quả: Khối lượng trung bình của từng loại quả (Average Weight):")
print(average_weight_by_fruit)
print("\n" + "="*40 + "\n")
weighted_sum_by_fruit = df.groupby('Fruit')['Weight'].sum()
print("Kiểm tra: Tổng khối lượng của từng loại quả (Weighted Sum):")
print(weighted_sum_by_fruit)
```

```python
# bài 4.21
import pandas as pd
data = {
    'country': [
        'Afghanistan', 'Albania', 'Algeria', 'Andorra', 'Angola', 
        'Antigua & Barbuda', 'Argentina', 'Armenia', 'Australia', 'Austria'
    ],
    'beer_servings': [
        0, 89, 25, 245, 217, 102, 29, 21, 261, 279
    ],
    'spirit_servings': [
        0, 132, 0, 138, 57, 128, 45, 179, 72, 75
    ],
    'wine_servings': [
        0, 54, 14, 312, 45, 56, 117, 11, 212, 191
    ],
    'total_litres_of_pure_alcohol': [
        0.0, 4.9, 0.7, 12.4, 5.9, 5.6, 8.3, 3.8, 10.4, 9.7
    ],
    'continent': [
        'Asia', 'Europe', 'Africa', 'Europe', 'Africa', 
        'North America', 'South America', 'Europe', 'Oceania', 'Europe'
    ]
}

drinks_df = pd.DataFrame(data)

drinks_df.to_csv('drinks.csv', index=False)

print("✅ Đã tạo file 'drinks.csv' thành công trong thư mục hiện tại.")
drink = pd.read_csv('drinks.csv', index_col=0)

print("\n--- Bài 4.21. Xử Lý Dataset drink.csv ---")
print(f"\nKiểu dữ liệu (type) của drink: {type(drink)}")
print(f"Kích thước (shape) của drink: {drink.shape}")
print("\nTên các cột (columns) của drink:")
print(drink.columns.tolist())
print("\n5 dòng đầu (head):")
print(drink.head())
print("\n5 dòng cuối (tail):")
print(drink.tail())
avg_beer_by_continent = drink.groupby('continent')['beer_servings'].mean()
print("\n2. Số lượng bia tiêu thụ trung bình ở mỗi châu lục:")
print(avg_beer_by_continent)

wine_stats_by_continent = drink.groupby('continent')['wine_servings'].describe()
print("\n3. Thông tin thống kê rượu vang (wine_servings) ở mỗi châu lục:")
print(wine_stats_by_continent)

avg_all_drinks_by_continent = drink.groupby('continent')[['beer_servings', 'spirit_servings', 'wine_servings']].mean()
print("\n4. Số lượng bia và rượu tiêu thụ trung bình ở mỗi châu lục:")
print(avg_all_drinks_by_continent)
median_all_drinks_by_continent = drink.groupby('continent')[['beer_servings', 'spirit_servings', 'wine_servings']].median()
print("\n5. Giá trị trung vị (median) của bia và rượu ở mỗi châu lục:")
print(median_all_drinks_by_continent)

spirit_metrics = drink.groupby('continent')['spirit_servings'].agg(['mean', 'max', 'min'])
print("\n6. Số lượng rượu mạnh (spirit_servings) trung bình, lớn nhất, nhỏ nhất ở mỗi châu lục:")
print(spirit_metrics)

drink_sorted_by_beer = drink.sort_values(by='beer_servings', ascending=True)

print("\n7. 5 quốc gia có lượng tiêu thụ bia ít nhất:")
print(drink_sorted_by_beer.head(5)['beer_servings'])

print("\n5 quốc gia có lượng tiêu thụ bia nhiều nhất:")
print(drink_sorted_by_beer.tail(5)['beer_servings'])
```

```python
#bài 5.1
import sqlite3

conn = sqlite3.connect("test.db")
cursor = conn.cursor()

cursor.execute("SELECT sqlite_version();")
version = cursor.fetchone()

print("SQLite version:", version[0])

conn.close()
```

```python
# bài 5.3
import sqlite3
conn = sqlite3.connect("school.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    age INTEGER
)
""")

print("Tạo bảng thành công")

conn.commit()
conn.close()
```

```python
# bài 5.4
import sqlite3

conn = sqlite3.connect("school.db")
cursor = conn.cursor()

cursor.execute("""
SELECT name FROM sqlite_master 
WHERE type='table';
""")

tables = cursor.fetchall()
print("Danh sách bảng:")
for table in tables:
    print(table[0])

conn.close()
```

```python
# bài 5.5
import sqlite3

conn = sqlite3.connect("school.db")
cursor = conn.cursor()

# Chèn dữ liệu
cursor.execute("INSERT INTO students (name, age) VALUES (?, ?)", ("An", 18))
cursor.execute("INSERT INTO students (name, age) VALUES (?, ?)", ("Bình", 19))
cursor.execute("INSERT INTO students (name, age) VALUES (?, ?)", ("Chi", 20))

conn.commit()

# Lấy dữ liệu
cursor.execute("SELECT * FROM students")
rows = cursor.fetchall()

print("Danh sách sinh viên:")
for row in rows:
    print(row)

conn.close()
```

```python
# bài 5.6
import sqlite3

conn = sqlite3.connect("school.db")
cursor = conn.cursor()

cursor.execute("SELECT COUNT(*) FROM students")
count = cursor.fetchone()

print("Số bản ghi:", count[0])

conn.close()
```

```python
class Date:
    def __init__(self, day=1, month=1, year=2000):
        self.day = day
        self.month = month
        self.year = year
        self._validate_date()
    
    def _is_leap_year(self):
        """Kiểm tra năm nhuận"""
        return (self.year % 4 == 0 and self.year % 100 != 0) or (self.year % 400 == 0)
    
    def _days_in_month(self):
        """Trả về số ngày trong tháng"""
        days_in_month = [31, 29 if self._is_leap_year() else 28, 31, 30, 31, 30,
                         31, 31, 30, 31, 30, 31]
        return days_in_month[self.month - 1]
    
    def _validate_date(self):
        """Kiểm tra và điều chỉnh ngày/tháng hợp lệ"""
        if self.month < 1 or self.month > 12:
            self.month = 1
        max_days = self._days_in_month()
        if self.day < 1 or self.day > max_days:
            self.day = 1
    
    def display(self):
        """In ra ngày/tháng/năm"""
        print(f"{self.day:02d}/{self.month:02d}/{self.year}")
    
    def next(self):
        """Tính ngày kế tiếp"""
        max_days = self._days_in_month()
        
        if self.day < max_days:
            self.day += 1
        else:
            self.day = 1
            if self.month < 12:
                self.month += 1
            else:
                self.month = 1
                self.year += 1
        return Date(self.day, self.month, self.year)


# ===== PHẦN CHẠY THỬ =====
print("\n=== TEST DATE ===")
# Nhập theo định dạng: dd mm yyyy
day, month, year = map(int, input("Nhập ngày tháng năm (cách nhau bằng khoảng trắng): ").split())

# Tạo đối tượng Date
date1 = Date(day, month, year)

print("Ngày hiện tại:")
date1.display()

print("Ngày tiếp theo:")
date1.next().display()
```

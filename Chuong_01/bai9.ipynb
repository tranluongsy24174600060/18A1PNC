```python
import math

# =================== LỚP CƠ SỞ: ĐA GIÁC ===================
class DaGiac:
    def __init__(self, so_canh: int, do_dai_canh: float):
        self.so_canh = so_canh
        self.do_dai_canh = do_dai_canh

    def tinh_chu_vi(self):
        return self.so_canh * self.do_dai_canh

    def tinh_dien_tich(self):
        """Tính diện tích đa giác đều"""
        if self.so_canh < 3:
            return 0
        return (self.so_canh * self.do_dai_canh**2) / (4 * math.tan(math.pi / self.so_canh))

    def __str__(self):
        return f"Đa giác {self.so_canh} cạnh: Chu vi = {self.tinh_chu_vi():.2f}, Diện tích = {self.tinh_dien_tich():.2f}"


# =================== LỚP HÌNH BÌNH HÀNH ===================
class HinhBinhHanh:
    def __init__(self, canh_day: float, canh_ben: float, goc: float):  # goc tính theo độ
        self.canh_day = canh_day
        self.canh_ben = canh_ben
        self.goc = math.radians(goc)  # chuyển sang radian để tính toán

    def tinh_chu_vi(self):
        return 2 * (self.canh_day + self.canh_ben)

    def tinh_dien_tich(self):
        return self.canh_day * self.canh_ben * math.sin(self.goc)

    def __str__(self):
        return f"Hình bình hành: Chu vi = {self.tinh_chu_vi():.2f}, Diện tích = {self.tinh_dien_tich():.2f}"


# =================== LỚP HÌNH CHỮ NHẬT ===================
class HinhChuNhat(HinhBinhHanh):
    def __init__(self, chieu_dai: float, chieu_rong: float):
        super().__init__(chieu_dai, chieu_rong, 90)  # góc vuông mặc định

    def tinh_dien_tich(self):
        return self.canh_day * self.canh_ben

    def __str__(self):
        return f"Hình chữ nhật: Chu vi = {self.tinh_chu_vi():.2f}, Diện tích = {self.tinh_dien_tich():.2f}"


# =================== LỚP HÌNH VUÔNG ===================
class HinhVuong(HinhChuNhat):
    def __init__(self, canh: float):
        super().__init__(canh, canh)

    def tinh_dien_tich(self):
        return self.canh_day ** 2

    def __str__(self):
        return f"Hình vuông: Chu vi = {self.tinh_chu_vi():.2f}, Diện tích = {self.tinh_dien_tich():.2f}"


# =================== CHƯƠNG TRÌNH CHÍNH ===================
def main():
    while True:
        print("\n========== MENU HÌNH HỌC ==========")
        print("1. Đa giác đều")
        print("2. Hình bình hành")
        print("3. Hình chữ nhật")
        print("4. Hình vuông")
        print("0. Thoát chương trình")
        lua_chon = input("👉 Mời bạn chọn hình cần tính: ")

        if lua_chon == "1":
            n = int(input("Nhập số cạnh của đa giác: "))
            a = float(input("Nhập độ dài mỗi cạnh: "))
            hinh = DaGiac(n, a)
        elif lua_chon == "2":
            a = float(input("Nhập độ dài cạnh đáy: "))
            b = float(input("Nhập độ dài cạnh bên: "))
            goc = float(input("Nhập góc giữa hai cạnh (độ): "))
            hinh = HinhBinhHanh(a, b, goc)
        elif lua_chon == "3":
            dai = float(input("Nhập chiều dài: "))
            rong = float(input("Nhập chiều rộng: "))
            hinh = HinhChuNhat(dai, rong)
        elif lua_chon == "4":
            canh = float(input("Nhập độ dài cạnh: "))
            hinh = HinhVuong(canh)
        elif lua_chon == "0":
            print("👋 Cảm ơn bạn đã sử dụng chương trình! Tạm biệt!")
            break
        else:
            print("⚠️ Lựa chọn không hợp lệ, vui lòng nhập lại.")
            continue

        print("➡ Kết quả:", hinh)


# =================== CHẠY CHƯƠNG TRÌNH ===================
if __name__ == "__main__":
    main()

```
